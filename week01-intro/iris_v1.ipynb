```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('Iris.csv')
```

```python
df['species_code'] = df['Species'].map({'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2})
```

# Prepare data for ML

```python
features = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']]
```

```python
target = df['species_code']
```

```python
trainX, testX, trainY, testY = train_test_split(features, target, test_size=0.2, random_state=42)
```

# Select Model, Train and Predict

```python
model = RandomForestClassifier()
```

```python
model.fit(trainX, trainY)
```

```python
prediction = model.predict(testX)
```

```python
accuracy = accuracy_score(testY, prediction)
```
