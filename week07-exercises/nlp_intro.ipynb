```python
reviews = [
    "The product is terrible. I hated every single feature of it!",
    "Amazing product, I loved it and recommend it to all my friends!",
    "Horrible",
    "Product is barely ok",
    "The package arrived on Tuesday at three in the afternoon and was left on the front porch"
]
```

```python
import nltk
```

```python
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
analyzer = SentimentIntensityAnalyzer()
```

```python
def check_polarity(text):
    scores = analyzer.polarity_scores(text)
    comp = scores['compound']
    if comp > 0.05:
        print(f"{comp} positive")
    elif comp < -0.05:
        print(f"{comp} negative")
    else:
        print(f"{comp} neutral")
```

```python
for review in reviews:
    check_polarity(review)
```
